## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Research Question**

Which content pages should be prioritized for review based on observable search performance, engagement, and freshness signals?

**Decision**

The project helps an SEO or content manager decide which pages should be reviewed first.

**Who acts and what do they do?**

An SEO/content manager reviews the ranked pages and decides whether to refresh, expand, rewrite, monitor, or leave the page unchanged.

**Cost of a wrong recommendation**

If low-priority pages are ranked too highly, reviewers spend time on pages with little opportunity. If high-priority declining pages are missed, traffic and engagement may continue to decline before they are reviewed.

**Why data/ML helps**

A simple rule such as "old pages first" cannot capture the interaction between visibility, engagement, freshness, CTR, and other signals. ML can combine multiple observable signals to produce a ranked review queue that better supports human decision-making. The goal is decision support, not automation.

## 3. Quick look at the data (2-3 real numbers)

The starter dataset contains **30,000 content pages**, **44 features**, and data from **32 unique clients**, providing a diverse set of observable search and engagement signals.

The distribution of `trend_direction` shows that **16,262 pages (about 54.2%)** are marked as **down**, indicating that a large portion of the content inventory may require review or monitoring.

These statistics suggest there are many established pages with measurable search performance, making the dataset suitable for building a content opportunity ranking system.
